```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def analyze_quantities(file_path):
    with open(file_path, 'r') as f:
        data = [list(map(float, line.strip().split())) for line in f if line.strip()]
    
    for idx, row in enumerate(data):
        row = np.array(row)
        mean = np.mean(row)
        std = np.std(row)
        
        print(f"Quantity {idx + 1}: mean = {mean:.3f}, std = {std:.3f}")
        
        # Plot histogram
        plt.figure(figsize=(6, 4))
        count, bins, _ = plt.hist(row, bins=15, density=True, alpha=0.6, color='skyblue', edgecolor='black', label='Data')
        
        # Overlay Gaussian
        x = np.linspace(min(row), max(row), 100)
        plt.plot(x, norm.pdf(x, mean, std), 'r-', label='Gaussian Fit')
        
        plt.title(f"Quantity {idx + 1}")
        plt.xlabel("Value")
        plt.ylabel("Density")
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.show()
```

```python
analyze_quantities( "c880_nets" )
```
