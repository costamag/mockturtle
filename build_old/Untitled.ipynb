```python
X = []
Y = []
with open("dac_02.txt", "r") as file1:
    for line in file1.readlines():
        f_list = [float(i) for i in line.split(" ")]
        print(f_list)
        X.append( f_list[0] )
        Y.append( f_list[1] )
```

```python
import matplotlib.pyplot as plt
plt.scatter( X, Y )
```

```python
from scipy import stats
res = stats.pearsonr(X, Y)
print(res)
```

```python
X = []
Y = []
with open("dac_02_model01.txt", "r") as file1:
    for line in file1.readlines():
        f_list = [float(i) for i in line.split(" ")]
        print(f_list)
        if( f_list[0] > 0 and f_list[1] > 0 ):
            X.append( f_list[0] )
            Y.append( f_list[1] )

```
