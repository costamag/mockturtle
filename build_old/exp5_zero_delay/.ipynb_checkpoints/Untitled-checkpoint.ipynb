```python
import numpy as np

def extract_data( file, min_arrival, max_arrival ):
    X = []
    Y = []
    with open( file, "r") as file1:
        for line in file1.readlines():
            f_list = [float(i) for i in line.split(" ")]
            X.append( [f_list[i]+0.01 for i in range(0, len(f_list)-2)] )
            Y.append( f_list[-2] )
    return [X,Y]
```
