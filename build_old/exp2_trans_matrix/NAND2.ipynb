```python
import numpy as np

def extract_data( file ):
    X = []
    Y = []
    with open( file, "r") as file1:
        for line in file1.readlines():
            f_list = [float(i) for i in line.split(" ")]
            #print(f_list)
            found_one = True
            for p in f_list:
                if p > 0.00001:
                    found_one = True
            for p in f_list:
                if p > 1:
                    found_one = False
            if( found_one ):#and ( f_list[-1] > 0.001 )):
                X.append( [f_list[i]+0.01 for i in range(0, len(f_list)-1)] )
                Y.append( f_list[-1] )
    return [X,Y]
```

```python
[X, Y] = extract_data("data_NAND2x1_ASAP7_75t_R.txt")
```

```python
threshold = round(len(X) * 0.8 )

X_train = np.array([X[i] for i in range( threshold ) ])
Y_train = np.array([Y[i] for i in range( threshold ) ])
X_test = np.array([X[i] for i in range( threshold, len(X) ) ])
Y_test = np.array([Y[i] for i in range( threshold, len(Y) ) ])
```
